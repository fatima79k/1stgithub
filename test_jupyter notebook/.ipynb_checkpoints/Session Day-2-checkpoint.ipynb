```python
# weight in kg/height in m and its square
```

```python
height = input("What is your height?")
```

```python
# type conversion
height = float(height)
```

```python
weight = input("What is your weight?")
```

```python
weight = float(weight)
```

```python
name = input("What is your name?")
```

```python
BMI = weight/height**2
BMI
```

```python
print("My name is", name, "and my BMI is", BMI)
```
