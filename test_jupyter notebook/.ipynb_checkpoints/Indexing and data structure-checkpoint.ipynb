## 1- Tuple
- ordered collection of elements
- enclosed in ( ) round braces/paranthesis
- different kind of elements can be stored
- once elements are stored, you cannot change them (unmutatable)

```python
# Types of elements
# integer, string, Boolean operator, float number
tup1 = (1, "python", True, 2.5)
tup1
```

```python
# type of a tuple
type(tup1)
```

# Indexing in tuple

```python
tup1[0]
```

```python
tup1[1]
```

```python
tup1[2]
```

```python
tup1[3]
```

```python
# last element is exclusive
tup1[0:3]
```

```python
# count of elements in tuple
len(tup1)
```

```python
tup2 = (2, "Ammar", 3.5, False)
tup2
```

```python
# Concatinate (to add two tuple or >2)
tup1 + tup2
```

```python
# Contatinate + Repeat
tup1*2 + tup2
```

```python
tup3 = (20, 50, 30, 20, 60, 79, 85)
tup3
```

```python
# Minimum
min(tup3)
```

```python
# Maximum
max(tup3)
```

```python
# Multiplication
tup3*2
```

```python
# Count how many times a value appears in tuple
tup3.count(20)
```

```python
# Find items position
tup3.index(30)
```

## 3. Dictionaries 
- An unordered collection of elements
- Key and value
- Curly braces or brackets { }
- Mutateable/change the values

```python
# food and their prices
food1 = { "Samosa":30, "Raita":20, "Pakora":100, "Salad":50, "Chicken Rolls":30}
food1
```

```python
type(food1)
```

```python
# extract date
keys1 = food1.keys()
keys1
```

```python
# All values
food1.values()
```

```python
# adding new elements 
food1["Tikki"]=10
food1
```

```python
# update the values 
food1["Tikki"]=15
food1
```

```python
food2 = {"Dates":50, "Chocolate":200, "Sawayyan":1000}
food2
```

```python
# Concatenate (add)
# Adds new key value pairs or changes existing values
food1.update(food2)
food1
```

```python
# remove all items
food1.clear()
food1
```

```python
# Creates a copy of of the dictionary
food1 = food2.copy()
food1
```

```python
# creates a dictionary using given keys with the same value
food1 = dict.fromkeys(keys1, 0)
food1
```

```python
# Get the value of specific key
food1.get("Samosa")
```

```python
# Return all key value pairs
food1.items()
```

```python
# Return all keys
food1.keys()
```

```python
# Remove a specified key 
food1.pop("Samosa")
food1
```

```python
# Remove and returns the last inserted key value pair
food1.popitem()
```

```python
# Returns the value if the key exists. If it does not exist, it adds the key with a default value
food1.setdefault("Pakora", 100)
food1
```

## 4- Sets
- Unordered and unindexed
- Curly braces are used { }
- No duplicates allowed 
- Boolean operators cannot be added

```python
s1 = {1, 2.2, 5.2, "Ammar", "Codanics", "Faisalabad", True}
s1
```

```python
# Add one element to a set
s1.add("Ammar1")
s1
```

```python
# Remove a specific elements
s1.remove("Ammar1")
s1
```

```python
# Remove all elements
s1.clear()
s1
```

```python
# Creates a copy of a set
s1.copy()
s1
```

```python
s2 = {1, 7, 9.2, "lahore", "Codanics", "python"}
s2
```

```python
# Return elements that are in the first set but not in the second
s1.difference(s2)
```

```python
# Remove elements that are also present in another set
s1.difference_update(s2)
s1
```

```python
# Remove a specific elements. If it does not exist, no error occurs.
s1.discard("Ammar")
s1
```

```python
# Return common elements
s1.intersection(s2)
```

```python
# Keeps only common elements in the first set
s1.intersection_update(s2)
s1
```

```python
# Checks whether two sets have no common elements
s1.isdisjoint(s2)
```

```python
# Checks whether all elements of one set are present in another
s1.issubset(s2)
```

```python
# Checks whether one set contains all elements of another set
s1.issuperset(s2)
```

```python
# Removes and returns an arbitrary elements
s1.pop()
```

```python
# Returns elements that are in either set, ut not in both
s1.symmetric_difference(s2)
```

```python
# Updates the first set with the symmetric difference
s1.symmetric_difference_update(s2)
s1
```

```python
# Combines two sets and remove duplicates
s1.union(s2)
```

```python
# Adds multiple elemets from another set
s1.update(s2)
s1
```
